```python
date_map = {
    "Feb20": (2020, 1),
    "May19": (2019, 2),
    "Oct19": (2019, 4),
    "Sep18": (2018, 3)
}
 #date_map[x] Looks up that date in your dictionary. For "Feb20", it returns (2020, 1)
 #[0] Selects the first item in the tuple. Python starts counting at zero, so (2020, 1)[0] gives 2020
df["year"] = df["date"].map(lambda x: date_map[x][0])           
df["quarter"] = df["date"].map(lambda x: date_map[x][1])
```

```python
df[["date", "year", "quarter"]].drop_duplicates()
```

```python
d = df[df["city"].isin(set(citys) - {"Greix"})].copy()
```

```python
then = gq[["City", "Year", "Quarter", "Index"]].rename(
    columns={"City": "city", "Year": "year", "Quarter": "quarter", "Index": "idx_then"})
```

```python
d = d.merge(then, on=["city", "year", "quarter"], how="left")
```

```python
print(len(d), d["idx_then"].isna().sum())
```

```python
d.tail(10)
```

```python
now = gq[(gq.Year == 2026) & (gq.Quarter == 2)].set_index("City")["Index"]
```

```python
d["idx_now"] = d["city"].map(now)
d["factor"] = d["idx_now"] / d["idx_then"]
d["factor"].describe()
```

```python
d.groupby("city")["factor"].mean().sort_values().round(2)
```

```python
d[d.city == "Berlin"]["factor"].mean()
```

```python
d["rent_sqm"] = d["baseRent"] / d["livingSpace"]
d[["baseRent", "livingSpace", "rent_sqm"]].describe()
```

```python
print((d.livingSpace < 10).sum(), (d.livingSpace > 300).sum())
print((d.baseRent < 100).sum(), (d.baseRent > 5000).sum())
print((d.rent_sqm < 2).sum(), (d.rent_sqm > 50).sum())
```

```python
clean = d[d.livingSpace.between(10, 300) & d.baseRent.between(100, 5000) & d.rent_sqm.between(2, 50)]
```

```python
print (len(clean), len(d) , len(d) - len (clean))
```

```python
print(clean.head(10).to_string())
```

```python
cols = ["city", "livingSpace", "noRooms", "yearConstructed", "balcony", "floor", "hasKitchen",
        "lift", "garden", "cellar", "newlyConst", "typeOfFlat", "condition", "heatingType"]
(clean[cols].isna().mean() * 100).round(1).sort_values(ascending=False)
```

```python
clean[["yearConstructed", "floor"]].describe()
```

```python
print((clean.yearConstructed < 1800).sum(), (clean.yearConstructed > 2022).sum(), (clean.floor > 30).sum())
```

```python
clean.loc[(clean.yearConstructed < 1800) | (clean.yearConstructed > 2022), "yearConstructed"] = None
clean.loc[clean.floor > 30, "floor"] = None

clean["year_missing"] = clean["yearConstructed"].isna()
clean["floor_missing"] = clean["floor"].isna()
clean[["year_missing", "floor_missing"]].mean().round(3)
```

```python
clean["yearConstructed"] = clean["yearConstructed"].fillna(clean["yearConstructed"].median())
clean["floor"] = clean["floor"].fillna(clean["floor"].median())
clean["typeOfFlat"] = clean["typeOfFlat"].fillna("unknown")
clean[["yearConstructed", "floor", "typeOfFlat"]].isna().sum()
```

```python
clean["rent_2026"] = (clean["baseRent"] * clean["factor"]).round(2)
clean[["baseRent", "rent_2026"]].describe()
```

```python
cols = ["city", "regio1", "date", "livingSpace", "noRooms", "yearConstructed", "floor",
        "balcony", "hasKitchen", "lift", "garden", "cellar", "newlyConst", "typeOfFlat",
        "year_missing", "floor_missing", "baseRent", "factor", "rent_2026"]
final = clean[cols].copy()
final.shape
```

```python
import sqlite3
Path("../data/processed").mkdir(parents=True, exist_ok=True)

con = sqlite3.connect("../data/processed/rent.db")
final.to_sql("listings", con, if_exists="replace", index=False)
pd.read_sql("SELECT COUNT(*) AS n FROM listings", con)
```
