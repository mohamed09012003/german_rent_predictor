```python
date_map = {
    "Feb20": (2020, 1),
    "May19": (2019, 2),
    "Oct19": (2019, 4),
    "Sep18": (2018, 3)
}
 #date_map[x] Looks up that date in your dictionary. For "Feb20", it returns (2020, 1)
 #[0] Selects the first item in the tuple. Python starts counting at zero, so (2020, 1)[0] gives 2020
df["year"] = df["date"].map(lambda x: date_map[x][0])           
df["quarter"] = df["date"].map(lambda x: date_map[x][1])
```

```python
df[["date", "year", "quarter"]].drop_duplicates()
```

```python
d = df[df["city"].isin(set(citys) - {"Greix"})].copy()
```

```python
then = gq[["City", "Year", "Quarter", "Index"]].rename(
    columns={"City": "city", "Year": "year", "Quarter": "quarter", "Index": "idx_then"})
```

```python
d = d.merge(then, on=["city", "year", "quarter"], how="left")
```

```python
print(len(d), d["idx_then"].isna().sum())
```

```python
d.tail(10)
```

```python
now = gq[(gq.Year == 2026) & (gq.Quarter == 2)].set_index("City")["Index"]
```

```python
d["idx_now"] = d["city"].map(now)
d["factor"] = d["idx_now"] / d["idx_then"]
d["factor"].describe()
```

```python
d.groupby("city")["factor"].mean().sort_values().round(2)
```
